```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# Define vocabulary size and sequence length for dummy data
vocab_size = 1000  # Example vocabulary size
seq_length = 10    # Example sequence length

# Generate dummy training data
num_samples = 1000

X_train = np.random.randint(0, vocab_size, size=(num_samples, seq_length))
# For text generation, y_train usually represents the next word (one-hot encoded)
y_train_labels = np.random.randint(0, vocab_size, size=(num_samples,))
y_train = to_categorical(y_train_labels, num_classes=vocab_size)

# Generate dummy validation data
num_val_samples = 200

X_val = np.random.randint(0, vocab_size, size=(num_val_samples, seq_length))
y_val_labels = np.random.randint(0, vocab_size, size=(num_val_samples,))
y_val = to_categorical(y_val_labels, num_classes=vocab_size)

print(f"Dummy data created:")
print(f"  vocab_size: {vocab_size}")
print(f"  seq_length: {seq_length}")
print(f"  X_train shape: {X_train.shape}")
print(f"  y_train shape: {y_train.shape}")
print(f"  X_val shape: {X_val.shape}")
print(f"  y_val shape: {y_val.shape}")
```
