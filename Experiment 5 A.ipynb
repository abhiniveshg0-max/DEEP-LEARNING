```python
# ============================================================
# EXPT NO: 5 - PART A: DATASET PREPARATION
# Student Roll No: 24BAD001
# ============================================================

# ---- Cell 1: Install & Import Libraries ----
!pip install datasets tensorflow numpy matplotlib -q

import numpy as np
import matplotlib.pyplot as plt
# from datasets import load_dataset # No longer needed for direct download
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical

# ---- Cell 2: Load the Tiny Shakespeare Dataset from an external URL ----
# The 'datasets' library no longer supports loading this dataset via script.
# We will download the raw text file directly.
import requests

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
raw_text = requests.get(url).text
print("Total characters in raw text:", len(raw_text))

# ---- Cell 3: Explore the Dataset ----
print("Sample of the text:\n")
print(raw_text[:500])

# ---- Cell 4: Convert Text to Lowercase ----
text = raw_text.lower()
print(text[:500])

# ---- Cell 5: Tokenize the Text into Individual Words ----
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

vocab_size = len(tokenizer.word_index) + 1  # +1 for padding token (index 0)
print("Vocabulary Size:", vocab_size)

# Show a few word -> index mappings
sample_words = list(tokenizer.word_index.items())[:10]
print("Sample word-to-index mappings:", sample_words)

# ---- Cell 6: Build Vocabulary and Generate Input Sequences for Next-Word Prediction ----
# Convert the whole text into a sequence of token indices
token_list = tokenizer.texts_to_sequences([text])[0]
print("Total tokens in corpus:", len(token_list))

# To keep training feasible, use a manageable subset of the corpus
token_list = token_list[:20000]

# Build n-gram sequences: each sequence is a growing prefix ending at the next word
input_sequences = []
seq_length = 10  # number of previous words used as context (window size)

for i in range(seq_length, len(token_list)):
    seq = token_list[i - seq_length:i + 1]  # 10 context words + 1 target word
    input_sequences.append(seq)

print("Total input sequences generated:", len(input_sequences))
print("Example sequence:", input_sequences[0])

# ---- Cell 7: Apply Sequence Padding ----
input_sequences = np.array(pad_sequences(
    input_sequences, maxlen=seq_length + 1, padding='pre'
))

# Split each sequence into input (X) and target/next word (y)
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# One-hot encode the target word for Softmax output
y = to_categorical(y, num_classes=vocab_size)

print("X shape:", X.shape)
print("y shape:", y.shape)

# ---- Cell 8: Split into Training and Validation Datasets ----
split_index = int(len(X) * 0.9)

X_train, X_val = X[:split_index], X[split_index:]
y_train, y_val = y[:split_index], y[split_index:]

print("Training samples:", X_train.shape[0])
print("Validation samples:", X_val.shape[0])
```
