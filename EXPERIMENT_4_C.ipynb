```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define a simple sequential model for demonstration purposes
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(10,)), # Example input shape
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid') # Example output layer for binary classification
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

Please note that `input_shape`, `loss`, and `metrics` should be adjusted according to your specific dataset and task. You will also need to define `train_ds`, `val_ds`, and `test_ds` (your training, validation, and test datasets, respectively) before running the `model.fit` and `model.evaluate` calls.

```python
import numpy as np

# --- Dummy Data Generation (Replace with your actual data) ---
# Assuming your model expects input_shape=(10,) and a single output
num_samples = 1000
num_features = 10

# Generate random features (e.g., for 10 input features)
x_data = np.random.rand(num_samples, num_features).astype(np.float32)
# Generate random labels (e.g., for binary classification)
y_data = (np.random.rand(num_samples) > 0.5).astype(np.float32)

# Split data into training and validation sets
train_split = int(0.8 * num_samples)
x_train, x_val = x_data[:train_split], x_data[train_split:]
y_train, y_val = y_data[:train_split], y_data[train_split:]

# --- Define tf.data.Dataset objects ---
# Batch size for training and validation
BATCH_SIZE = 32

# Create training dataset
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train))
train_ds = train_ds.shuffle(buffer_size=1024).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# Create validation dataset
val_ds = tf.data.Dataset.from_tensor_slices((x_val, y_val))
val_ds = val_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# If you also have a test_ds, you would define it similarly:
# x_test = np.random.rand(100, num_features).astype(np.float32)
# y_test = (np.random.rand(100) > 0.5).astype(np.float32)
# test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test))
# test_ds = test_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

print(f"train_ds created with {len(x_train)} samples")
print(f"val_ds created with {len(x_val)} samples")
```

### Explanation of the Dataset Creation

1.  **Dummy Data Generation**: This section creates `x_data` (features) and `y_data` (labels) using NumPy. `num_features` is set to 10 to match the `input_shape=(10,)` defined in your model. You **must** replace this with your actual data loading and preprocessing.
2.  **Splitting Data**: The data is split into training (`x_train`, `y_train`) and validation (`x_val`, `y_val`) sets.
3.  **`tf.data.Dataset.from_tensor_slices`**: This function is used to create a `tf.data.Dataset` from your NumPy arrays. Each element of the dataset will be a pair `(feature, label)`.
4.  **`shuffle(buffer_size)`**: This shuffles the training data to ensure randomness during training, which helps in preventing the model from learning the order of samples.
5.  **`batch(BATCH_SIZE)`**: This groups elements of the dataset into batches of a specified size. Training is typically done in batches.
6.  **`prefetch(tf.data.AUTOTUNE)`**: This allows the dataset to fetch batches in the background while the model is training on the current batch, significantly improving performance.

```python
import matplotlib.pyplot as plt

EPOCHS = 10

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)

train_acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
train_loss = history.history['loss']
val_loss = history.history['val_loss']

print("\nFinal epoch results:")
print(f"Training accuracy   : {train_acc[-1]:.4f}")
print(f"Validation accuracy : {val_acc[-1]:.4f}")
print(f"Training loss       : {train_loss[-1]:.4f}")
print(f"Validation loss     : {val_loss[-1]:.4f}")

test_loss, test_acc = model.evaluate(test_ds, verbose=0)
print(f"Test accuracy        : {test_acc:.4f}")
print(f"Test loss            : {test_loss:.4f}")

epochs_range = range(1, EPOCHS + 1)

plt.figure(figsize=(8, 5))
plt.plot(epochs_range, train_acc, label='Training accuracy')
plt.plot(epochs_range, val_acc, label='Validation accuracy')
plt.title('Accuracy vs Epoch')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(epochs_range, train_loss, label='Training loss')
plt.plot(epochs_range, val_loss, label='Validation loss')
plt.title('Loss vs Epoch')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```
