```python
# =============================================================
# PART B: CNN MODEL IMPLEMENTATION
# CNN for Image Classification — CIFAR-10
# =============================================================

import tensorflow as tf
from tensorflow.keras import layers, models

NUM_CLASSES = 10          # CIFAR-10 has 10 classes
INPUT_SHAPE = (32, 32, 3) # from Part A preprocessing

# -------------------------------------------------------------
# Build the CNN model
# -------------------------------------------------------------
model = models.Sequential([
    layers.Input(shape=INPUT_SHAPE),

    # --- Convolution Block 1 ---
    layers.Conv2D(filters=32, kernel_size=(3, 3), padding='same'),
    layers.Activation('relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),          # 32x32 -> 16x16

    # --- Convolution Block 2 ---
    layers.Conv2D(filters=64, kernel_size=(3, 3), padding='same'),
    layers.Activation('relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),          # 16x16 -> 8x8

    # --- Convolution Block 3 ---
    layers.Conv2D(filters=128, kernel_size=(3, 3), padding='same'),
    layers.Activation('relu'),
    layers.MaxPooling2D(pool_size=(2, 2)),          # 8x8 -> 4x4

    # --- Flatten + Fully Connected layers ---
    layers.Flatten(),
    layers.Dense(128),
    layers.Activation('relu'),
    layers.Dropout(0.5),                            # reduces overfitting

    # --- Softmax output layer ---
    layers.Dense(NUM_CLASSES),
    layers.Activation('softmax')
])

# -------------------------------------------------------------
# Compile the model
# -------------------------------------------------------------
model.compile(
    optimizer='adam',                                  # appropriate optimizer
    loss='sparse_categorical_crossentropy',             # cross-entropy loss
    metrics=['accuracy']
)

# -------------------------------------------------------------
# Model summary
# -------------------------------------------------------------
model.summary()

# model is now ready to be trained in Part C using:
# history = model.fit(x_train, y_train, validation_data=(x_val, y_val), epochs=..., batch_size=...)
```
