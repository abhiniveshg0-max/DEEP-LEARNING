```python
# Roll No: 24BAD001
# Experiment 2 - Visualization of Activation Functions

import numpy as np
import matplotlib.pyplot as plt

# Generate input values
x = np.linspace(-10, 10, 200)

# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh Function
def tanh(x):
    return np.tanh(x)

# ReLU Function
def relu(x):
    return np.maximum(0, x)

# Plot all activation functions
plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid(x), label='Sigmoid', linewidth=2)
plt.plot(x, tanh(x), label='Tanh', linewidth=2)
plt.plot(x, relu(x), label='ReLU', linewidth=2)

# Graph details
plt.title("Visualization of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

# Display graph
plt.show()
```
