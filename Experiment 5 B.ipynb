```python
# ============================================================
# EXPT NO: 5 - PART B: IMPLEMENTING THE RNN MODEL
# Student Roll No: 24BAD001
# ============================================================

# ---- Cell 1: Import Libraries ----
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.optimizers import Adam

# NOTE: This continues from Part A.
# vocab_size, seq_length, X_train, y_train, X_val, y_val must already exist
# from the Part A dataset preparation code.

# ---- Cell 2: Define Model Hyperparameters ----
# Placeholder for seq_length - replace with actual value from Part A
# For demonstration, setting a default value. In a real scenario, this should be dynamically loaded.
seq_length = 10 # Example placeholder value, adjust as per your Part A data

# Placeholder for vocab_size - replace with actual value from Part A
# For demonstration, setting a default value. In a real scenario, this should be dynamically loaded.
vocab_size = 10000 # Example placeholder value, adjust as per your Part A data

embedding_dim = 100   # size of the dense word vectors
rnn_units = 150        # number of units in the SimpleRNN layer
input_length = seq_length  # length of input sequences (from Part A)

# ---- Cell 3: Build the RNN Model ----
model = Sequential()

# Embedding layer: converts word indices into dense vector representations
model.add(Embedding(input_dim=vocab_size,
                     output_dim=embedding_dim))

# Simple RNN layer: learns sequential dependencies in the text
model.add(SimpleRNN(rnn_units, return_sequences=False))

# Dense output layer with Softmax activation for next-word prediction
model.add(Dense(vocab_size, activation='softmax'))

# ---- Cell 4: Compile the Model ----
model.compile(
    loss='categorical_crossentropy',
    optimizer=Adam(learning_rate=0.001),
    metrics=['accuracy']
)

# ---- Cell 5: Display Model Architecture and Summary ----
model.build(input_shape=(None, input_length))
model.summary()
```
