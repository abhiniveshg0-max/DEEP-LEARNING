```python
print("24BAD001")

!pip -q install tensorflow transformers


import os
import zipfile
import random
import shutil
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from PIL import Image

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices("GPU"))


# 1. UPLOAD DATASET


print("\nUpload your Flowers Recognition ZIP file")
uploaded = files.upload()

zip_files = [f for f in uploaded.keys() if f.lower().endswith(".zip")]

if not zip_files:
    raise Exception("Please upload the Flowers Recognition ZIP file.")

zip_path = zip_files[0]

print("Uploaded:", zip_path)


# 2. EXTRACT DATASET
extract_dir = "/content/flowers_data"
if os.path.exists(extract_dir):
    shutil.rmtree(extract_dir)
os.makedirs(extract_dir)
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_dir)
print("\nDataset extracted successfully.")

# 3. FIND DATASET FOLDER AUTOMATICALLY

data_dir = None
for root, dirs, files_list in os.walk(extract_dir):
    # Ignore empty directories
    if not dirs:
        continue
    valid_class_dirs = 0
    for d in dirs:
        folder = os.path.join(root, d)
        try:
            image_files = [
                f for f in os.listdir(folder)
                if f.lower().endswith(
                    (".jpg", ".jpeg", ".png", ".bmp", ".gif")
                )
            ]
            if len(image_files) > 0:
                valid_class_dirs += 1
        except:
            pass
    if valid_class_dirs >= 2:
        data_dir = root
        break
if data_dir is None:
    raise Exception(
        "Could not automatically find the dataset folder. "
        "Please check the ZIP structure."
    )
print("\nDataset folder found:")
print(data_dir)

# 4. DISPLAY CLASS INFORMATION
class_folders = []
for folder in sorted(os.listdir(data_dir)):
    folder_path = os.path.join(data_dir, folder)
    if os.path.isdir(folder_path):
        image_count = len([
            f for f in os.listdir(folder_path)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png", ".bmp", ".gif") ) ])
        if image_count > 0:
            class_folders.append(folder)
            print(f"{folder}: {image_count} images")
NUM_CLASSES = len(class_folders)
print("\nNumber of classes:", NUM_CLASSES)
if NUM_CLASSES < 2:
    raise Exception("Dataset must contain at least 2 classes.")


# 5. IMAGE DATA GENERATOR


IMG_SIZE = 224
BATCH_SIZE = 32

datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    validation_split=0.30
)

train_generator = datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training",
    shuffle=True,
    seed=42
)

validation_generator = datagen.flow_from_directory(
    data_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation",
    shuffle=False,
    seed=42
)

print("\nClass indices:")
print(train_generator.class_indices)

print("\nTraining images:", train_generator.samples)
print("Validation/Test images:", validation_generator.samples)


# 6. DISPLAY SAMPLE IMAGES


images, labels = next(train_generator)

plt.figure(figsize=(10, 8))

for i in range(min(9, len(images))):

    plt.subplot(3, 3, i + 1)

    # Convert preprocessing approximately back for display
    img = images[i].copy()

    img = img - img.min()
    img = img / img.max()

    plt.imshow(img)
    plt.axis("off")

plt.suptitle("Sample Images from Flowers Dataset")
plt.show()


# 7. LOAD PRE-TRAINED RESNET50


print("\nLoading pre-trained ResNet50...")

base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Freeze pre-trained layers
for layer in base_model.layers:
    layer.trainable = False

print("ResNet50 loaded successfully.")
print("Pre-trained layers frozen.")


# 8. ADD NEW CLASSIFICATION HEAD


x = base_model.output

x = GlobalAveragePooling2D()(x)

x = Dense(
    128,
    activation="relu"
)(x)

output_layer = Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(
    inputs=base_model.input,
    outputs=output_layer
)


# 9. COMPILE MODEL


model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nModel Summary:")
model.summary()


# 10. TRAIN MODEL


EPOCHS = 10

print("\nStarting Transfer Learning Training...")

history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=EPOCHS
)


# 11. PRINT TRAINING RESULTS
print("\n================ TRAINING RESULTS ================")
for i in range(EPOCHS):

    print(
        f"Epoch {i+1}: "
        f"Train Accuracy = {history.history['accuracy'][i]:.4f}, "
        f"Validation Accuracy = {history.history['val_accuracy'][i]:.4f}, "
        f"Train Loss = {history.history['loss'][i]:.4f}, "
        f"Validation Loss = {history.history['val_loss'][i]:.4f}"
    )


# 12. ACCURACY GRAPH
plt.figure(figsize=(8, 5))-
plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid()

plt.show()


# 13. LOSS GRAPH


plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid()

plt.show()


# 14. MODEL EVALUATION


print("\n================ MODEL EVALUATION ================")

test_loss, test_accuracy = model.evaluate(
    validation_generator
)

print("Test/Validation Loss:", round(test_loss, 4))
print("Test/Validation Accuracy:", round(test_accuracy * 100, 2), "%")


# 15. HUGGING FACE VISION TRANSFORMER


print("\nLoading Hugging Face Vision Transformer...")

from transformers import pipeline

hf_model = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)

print("Hugging Face ViT loaded successfully.")


# 16. GET SAMPLE IMAGES FOR COMPARISON


sample_paths = validation_generator.filepaths

if len(sample_paths) > 5:
    sample_imgs = random.sample(sample_paths, 5)
else:
    sample_imgs = sample_paths

class_names = list(
    train_generator.class_indices.keys()
)


# 17. COMPARE RESNET50 AND HUGGING FACE ViT


print("\n================ MODEL COMPARISON ================")

for img_path in sample_imgs:

    image = Image.open(img_path).convert("RGB")



    resized = image.resize(
        (IMG_SIZE, IMG_SIZE)
    )

    img_array = np.array(resized)

    img_array = preprocess_input(
        img_array
    )

    img_array = np.expand_dims(
        img_array,
        axis=0
    )

    prediction = model.predict(
        img_array,
        verbose=0
    )

    predicted_index = np.argmax(prediction)

    resnet_class = class_names[
        predicted_index
    ]

    resnet_confidence = prediction[
        0
    ][predicted_index]



    hf_prediction = hf_model(image)

    hf_label = hf_prediction[0]["label"]
    hf_confidence = hf_prediction[0]["score"]

    print("\nImage:", os.path.basename(img_path))

    print(
        "Actual class:",
        os.path.basename(
            os.path.dirname(img_path)
        )
    )

    print(
        "ResNet50 Prediction:",
        resnet_class,
        "| Confidence:",
        round(resnet_confidence, 3)
    )

    print(
        "Hugging Face ViT Prediction:",
        hf_label,
        "| Confidence:",
        round(hf_confidence, 3)
    )

    print("-" * 60)


# 18. DISPLAY FINAL RESULT



print("              EXPERIMENT COMPLETED")


print("\nTransfer Learning Model: ResNet50")
print("Pre-trained Model: ImageNet")
print("Dataset: Flowers Recognition")
print("Number of Classes:", NUM_CLASSES)
print("Epochs:", EPOCHS)

print(
    "\nFinal Training Accuracy:",
    round(history.history["accuracy"][-1] * 100, 2),
    "%"
)

print(
    "Final Validation Accuracy:",
    round(history.history["val_accuracy"][-1] * 100, 2),
    "%"
)

print(
    "Test/Validation Accuracy:",
    round(test_accuracy * 100, 2),
    "%"
)

print("\nObservation:")
print(
    "The ResNet50 model uses pre-trained ImageNet features "
    "and a new classification layer trained for the flower classes. "
    "The Hugging Face ViT model is a general ImageNet classifier, "
    "so its labels may differ from the flower-specific classes."
)

print("\nExperiment 4 completed successfully!")
```
